```python
def task_calculate_average(num_list):
    """
    计算输入列表中所有数字的平均值
    参数:
        num_list (list): 包含数字的列表（如[1,2,3,4]）
    返回值:
        float: 列表中数字的平均值
    """
    if not num_list:
        return 0.0
    return sum(num_list) / len(num_list)
# demo测试代码
test_input = [1, 2, 3, 4]
avg_value = task_calculate_average(test_input)
print(avg_value)
# 运行函数
num_list = None  # 请根据实际需求赋值（包含数字的列表（如[1,2,3,4]））
avg_value = task_calculate_average(num_list)
print(f"输出结果（avg_value）：{avg_value}")
```
